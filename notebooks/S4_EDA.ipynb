This notebook verifies the cleaned NYC dataset and the POI data used by the React map.

Cells:
1. Load the cleaned CSV and inspect shape plus category counts.

```python
import pandas as pd

df = pd.read_csv('../data/processed/foursquare_nyc_clean.csv')
print(df.shape)
print(df['venue_category'].value_counts())
```

```python
import json

with open('../frontend/public/pois.json') as f:
    pois = json.load(f)

print(f"POIs: {len(pois)}")
print(pois[0])
```

```python
pip install folium
```

```python
import folium

m = folium.Map(location=[40.73, -73.93], zoom_start=11)
for poi in pois[:100]:
    folium.CircleMarker(
        [poi['lat'], poi['lng']],
        radius=4,
        popup=poi.get('category', 'Unknown'),
    ).add_to(m)

m.save('nyc_preview.html')
print('Map saved!')
```
